```python
import requests
import pandas as pd

BASE_URL = "https://api.fda.gov/device/event.json"

SEARCH = (
    'device.device_report_product_code:"LOX" '
    'AND date_received:[20250101 TO 20251231]'
)

response = requests.get(
    BASE_URL,
    params={"search": SEARCH, "limit": 100},
    timeout=60,
)
response.raise_for_status()

payload = response.json()
records = payload["results"]

print("Total matching records:", payload["meta"]["results"]["total"])
print("Sample records downloaded:", len(records))
```

```python
fields = [
    "mdr_report_key",
    "report_number",
    "date_received",
    "date_of_event",
    "event_type",
    "report_source_code",
    "product_problems",
]

sample = pd.DataFrame(records).reindex(columns=fields)

display(sample.head(10))
display(sample.isna().mean().rename("missing_fraction"))
```
